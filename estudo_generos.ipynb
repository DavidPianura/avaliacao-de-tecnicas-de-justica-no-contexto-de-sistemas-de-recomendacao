```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

## MOVIELENS

```python
path_100k = 'data/raw/movielens/100K/ml-100k/u.data'
path_g_100k = 'data/raw/movielens/100K/movies.dat'

ml = pd.read_csv(path_100k, sep='\t', names=['userId', 'movieId', 'rating', 'timestamp'], encoding='ISO-8859-1')
ml_g = pd.read_csv(path_g_100k, sep='::', names=['movieId', 'name', 'genres'], encoding='ISO-8859-1')
```

```python
all_genres = ml_g['genres'].str.split('|').explode()
```

```python
genre_counts = all_genres.value_counts().sort_values()
```

```python
plt.figure(figsize=(10, 6))
genre_counts.plot(kind='bar', color='grey', edgecolor='black')
plt.title('Frequência de Gêneros de Filmes')
plt.xlabel('Gêneros')
plt.ylabel('Frequência')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

```python
rating_frequency = ml['rating'].value_counts().sort_index()
rating_frequency
```

```python
plt.figure(figsize=(10, 6))
rating_frequency.plot(kind='bar', color='gray', edgecolor='black')
plt.title('Frequência de Ratings')
plt.xlabel('Avaliação')
plt.ylabel('Frequência')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```
